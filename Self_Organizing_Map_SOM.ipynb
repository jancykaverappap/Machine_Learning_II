```python
import numpy as np
import pandas as pd

url = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/iris.csv"
data = pd.read_csv(url)

features = ['sepal_length', 'sepal_width', 'petal_length', 'petal_width']
X = data[features].values
X = (X - X.min(axis=0)) / (X.max(axis=0) - X.min(axis=0))
X = np.round(X[:6], 2)

display(pd.DataFrame(X, columns=features))
```

```python
np.set_printoptions(precision=2, suppress=True)

n, f = X.shape

W = np.array([
    [0.2, 0.4, 0.6, 0.8],
    [0.7, 0.5, 0.3, 0.1]
])

epochs = 10
alpha = 0.5

for epoch in range(epochs):
    print("\nEpoch:", epoch + 1)
    for i in range(n):
        x = X[i]
        d1 = np.sqrt(np.sum((x - W[0]) ** 2))
        d2 = np.sqrt(np.sum((x - W[1]) ** 2))
        if d1 < d2:
            winner = 0
        else:
            winner = 1
        W[winner] = W[winner] + alpha * (x - W[winner])
        print("Input:", x)
        print("Distance 1:", round(d1, 2))
        print("Distance 2:", round(d2, 2))
        print("Winning neuron:", winner + 1)
        print("Updated weights:\n", W)

print("\nFinal Weight Matrix:")
print(W)

print("\nFinal Clusters:")
for i in range(n):
    x = X[i]
    d1 = np.sqrt(np.sum((x - W[0]) ** 2))
    d2 = np.sqrt(np.sum((x - W[1]) ** 2))
    cluster = 1 if d1 < d2 else 2
    print("Sample", i + 1, "belongs to neuron", cluster)
```
